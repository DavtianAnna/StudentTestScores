```python
X_train = df_train.drop('exam_score', axis=1)
y_train = df_train['exam_score']
X_test = df_test.copy()
```

```python
# SGDRegressor
```

```python
from sklearn.linear_model import SGDRegressor
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
model = SGDRegressor()
pipe = Pipeline([('scaler', scaler), ('model', model)])
```

```python
params = {
    'model__loss': ['squared_error', 'huber', 'epsilon_insensitive'],
    'model__penalty': ['l2', 'l1', 'elasticnet'],
    'model__alpha': [0.0001, 0.001, 0.01, 0.1],
    'model__learning_rate': ['constant', 'optimal', 'adaptive'],
    'model__eta0': [0.001, 0.01, 0.1],
    'model__max_iter': [1000, 2000, 3000]
}
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
grid_model = GridSearchCV(pipe, params, cv=3, verbose=2)
```

```python
grid_model.fit(X_train, y_train)
```

```python
y_pred = grid_model.predict(X_test)
df_sub = pd.read_csv('sample_submission.csv')
df_sub['exam_score'] = y_pred
df_sub.to_csv('submission1.csv', index=False, index_label=False)
```

```python
# XGBRegressor
```

```python
categorical_cols = [
    "gender",
    "course",
    "sleep_quality",
    "internet_access",
    "study_method",
    "facility_rating",
    "exam_difficulty",
]
```

```python
for i in categorical_cols:
    df_train[i] = label.fit_transform(df_train[i])
    df_test[i] = label.transform(df_test[i])
```

```python
X_train = df_train.drop(["exam_score", "id"], axis=1)
y_train = df_train["exam_score"]
X_test = df_test.drop(["id"], axis=1)
```

```python
from xgboost import XGBRegressor
from sklearn.pipeline import Pipeline
from sklearn.model_selection import GridSearchCV
```

```python
pipe = Pipeline([('model', XGBRegressor(random_state=42, n_jobs=-1))])
```

```python
params = {
    'model__n_estimators': [100, 300, 500],
    'model__learning_rate': [0.01, 0.05, 0.1],
    'model__max_depth': [3, 5, 7],
    'model__subsample': [0.8, 1.0]
}
```

```python
grid_model = GridSearchCV(pipe, params, cv=3, verbose=2, scoring='neg_root_mean_squared_error')
grid_model.fit(X_train, y_train)
```

```python
y_pred = grid_model.predict(X_test)
df_sub = pd.read_csv('sample_submission.csv')
df_sub['exam_score'] = y_pred
df_sub.to_csv('submission2.csv', index=False, index_label=False)
```
